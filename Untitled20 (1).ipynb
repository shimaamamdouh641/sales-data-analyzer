Data Visualization

```python
plt.plot(monthy_revenue)
plt.show()
```

```python
sales_df.groupby("Country")["revenue"].sum().sort_values(ascending=False).head(10)
```

```python
sns.boxplot(x=sales_df["revenue"])
plt.title("revenue distrubusion and outlier")
plt.show()
```

```python
top_products_revenue = (sales_df.groupby("Description")["revenue"].sum().sort_values(ascending=False).head(10))
sns.barplot(x=top_products_revenue.values,y=top_products_revenue.index)
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.show()
```

```python
top_products = (sales_df.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10))

top_products
```

```python
top_products = (sales_df.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10).reset_index())
sns.barplot(data=top_products,y="Description",x="Quantity")
plt.title("Top 10 Products by Quantity")
plt.show()
```

```python
top_countries = (sales_df.groupby("Country")["revenue"].sum().sort_values(ascending=False).head(10).reset_index())
sns.barplot(data=top_countries,y="Country",x="revenue")
plt.title("Top 10 Countries by Revenue")
plt.show()
```

```python
sns.histplot(data=sales_df, x="revenue", bins=20, kde=True)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Quantity")
plt.show()
```
